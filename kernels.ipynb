```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
beta = 9.0
sigma = 0.1
gamma = 0.0128

x = np.linspace(0.0, 1.0, 500)

f1 = x ** beta
f2 = np.exp(-(1 - x) / sigma)
f3 = np.exp(-((1 - x) ** 2) / gamma)

plt.figure(figsize=(8, 5))
plt.plot(x, f1, label=fr"$x^\beta$ (beta={beta})", linewidth=2)
plt.plot(x, f2, label=fr"$e^{{-(1-x)/\sigma}}$ (sigma={sigma})", linewidth=2)
plt.plot(x, f3, label=fr"$e^{{-(1-x)^2/\gamma}}$ (gamma={gamma})", linewidth=2)

plt.xlim(0, 1)
plt.ylim(0, 1.05)
plt.xlabel("c")
plt.ylabel("f(c)")
plt.title("Kernel Functions on [0, 1]")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
